```python
import numpy as np

def cohen_d(group1, group2):
    diff = np.mean(group1) - np.mean(group2)
    pooled_std = np.sqrt((np.var(group1, ddof=1) + np.var(group2, ddof=1)) / 2)
    return diff / pooled_std

d = cohen_d(exp_scores, ctrl_scores)
print(f"\nCohen's d = {d:.2f}")
print("Interpretation:")
if abs(d) < 0.2: print("   Small effect")
elif abs(d) < 0.5: print("   Medium effect")
else: print("   Large effect")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import FancyArrowPatch

# =========================
# Data
# =========================
countries = ['Bangladesh', 'Indonesia', 'Hungary']


# =========================
# Style variables
# =========================
AI_COLOR = "#FF6B6B"      # coral/red
CTRL_COLOR = "#4ECDC4"    # teal
EDGE = "black"
ALPHA = 0.85

TITLE_SIZE = 16
SUBTITLE_SIZE = 14
LABEL_SIZE = 12
TICK_SIZE = 11
ANNOT_SIZE = 10

# =========================
# Helper functions
# =========================
def add_bar_labels(ax, bars, fmt="{:.1f}", vpad=0.1):
    """Add value labels on bars with smart vertical padding."""
    for bar in bars:
        h = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2, h + vpad,
                fmt.format(h), ha="center", va="bottom",
                fontsize=ANNOT_SIZE, fontweight="bold")

def donut(ax, sizes, labels, colors, title):
    """Draw a donut pie with center hole & bold inside labels."""
    wedges, texts, autotexts = ax.pie(
        sizes, labels=labels, colors=colors,
        autopct="%1.1f", startangle=90,
        pctdistance=0.72, textprops={'fontsize': 10}
    )
    # center hole
    centre = plt.Circle((0, 0), 0.55, fc="white")
    ax.add_artist(centre)
    for t in autotexts:
        t.set_color("white")
        t.set_fontweight("bold")
    ax.set_title(title, fontsize=SUBTITLE_SIZE, pad=10)
    ax.axis("equal")

def add_paradox_arrows(ax, x, y_from, y_to, text, color):
    """Arrow annotation between two bars (AI vs Control)."""
    ax.add_patch(FancyArrowPatch(
        (x-0.2, y_from), (x+0.2, y_to),
        arrowstyle="->", mutation_scale=12,
        linewidth=2, color=color
    ))
    x_text = x + (0.42 if y_to >= y_from else 0.42)
    y_text = (y_from + y_to)/2
    ax.text(x_text, y_text, text, fontsize=ANNOT_SIZE,
            fontweight="bold", color=color,
            va="center", ha="left")

# =========================
# Derived stats
# =========================
x = np.arange(len(countries))
width = 0.36

avg_ai_effort = float(np.mean(ai_perceived_effort))
avg_ctrl_effort = float(np.mean(control_perceived_effort))
avg_ai_load = float(np.mean(ai_cognitive_load))
avg_ctrl_load = float(np.mean(control_cognitive_load))

effort_reduction = avg_ctrl_effort - avg_ai_effort
load_increase = avg_ai_load - avg_ctrl_load

# =========================
# Figure layout
# =========================
fig = plt.figure(figsize=(18, 13), constrained_layout=True)
gs = fig.add_gridspec(3, 3)

# 1) Main comparison: Perceived Effort (AI vs Control), grouped by country
ax1 = fig.add_subplot(gs[0, :2])
bars_ai = ax1.bar(x - width/2, ai_perceived_effort, width,
                  label="AI-Assisted", color=AI_COLOR, alpha=ALPHA,
                  edgecolor=EDGE, linewidth=0.6)
bars_ctrl = ax1.bar(x + width/2, control_perceived_effort, width,
                    label="Control", color=CTRL_COLOR, alpha=ALPHA,
                    edgecolor=EDGE, linewidth=0.6)

ax1.set_title("PERCEIVED EFFORT: AI vs Control (Lower is Better)", fontsize=TITLE_SIZE, pad=12)
ax1.set_xlabel("Countries", fontsize=LABEL_SIZE)
ax1.set_ylabel("Perceived Effort Score", fontsize=LABEL_SIZE)
ax1.set_xticks(x)
ax1.set_xticklabels(countries, fontsize=TICK_SIZE)
ax1.grid(True, axis="y", alpha=0.25)
ax1.set_ylim(0, max(control_perceived_effort) + 1.2)
ax1.legend(fontsize=11, frameon=False, ncols=2, loc="upper right")

add_bar_labels(ax1, bars_ai)
add_bar_labels(ax1, bars_ctrl)

# 2) Cognitive Load (AI vs Control), grouped by country
ax2 = fig.add_subplot(gs[0, 2])
bars_ai2 = ax2.bar(x - width/2, ai_cognitive_load, width,
                   label="AI-Assisted", color=AI_COLOR, alpha=ALPHA,
                   edgecolor=EDGE, linewidth=0.6)
bars_ctrl2 = ax2.bar(x + width/2, control_cognitive_load, width,
                     label="Control", color=CTRL_COLOR, alpha=ALPHA,
                     edgecolor=EDGE, linewidth=0.6)

ax2.set_title("COGNITIVE LOAD: AI vs Control (Higher = More Overwhelmed)", fontsize=TITLE_SIZE, pad=12)
ax2.set_xlabel("Countries", fontsize=LABEL_SIZE)
ax2.set_ylabel("Cognitive Load Score", fontsize=LABEL_SIZE)
ax2.set_xticks(x)
ax2.set_xticklabels(countries, fontsize=TICK_SIZE)
ax2.grid(True, axis="y", alpha=0.25)
ax2.set_ylim(0, max(ai_cognitive_load + control_cognitive_load) + 1.0)
ax2.legend(fontsize=10, frameon=False)

add_bar_labels(ax2, bars_ai2)
add_bar_labels(ax2, bars_ctrl2)

# 3) Donut pies: Average Perceived Effort (AI vs Control)
ax3 = fig.add_subplot(gs[1, 0])
donut(
    ax3,
    sizes=[avg_ai_effort, avg_ctrl_effort],
    labels=[f"AI\n{avg_ai_effort:.1f}", f"Control\n{avg_ctrl_effort:.1f}"],
    colors=[AI_COLOR, CTRL_COLOR],
    title="Average Perceived Effort (Lower with AI)"
)

# 4) Donut pies: Average Cognitive Load (AI vs Control)
ax4 = fig.add_subplot(gs[1, 1])
donut(
    ax4,
    sizes=[avg_ai_load, avg_ctrl_load],
    labels=[f"AI\n{avg_ai_load:.1f}", f"Control\n{avg_ctrl_load:.1f}"],
    colors=[AI_COLOR, CTRL_COLOR],
    title="Average Cognitive Load (Higher with AI)"
)

# 5) The AI Paradox (compact, 2 bars per metric)
ax5 = fig.add_subplot(gs[1:, 2])
categories = ["Perceived Effort", "Cognitive Load"]
ai_scores = [avg_ai_effort, avg_ai_load]
ctrl_scores = [avg_ctrl_effort, avg_ctrl_load]
x_paradox = np.arange(len(categories))

p_ai = ax5.bar(x_paradox - 0.2, ai_scores, 0.4, label="AI-Assisted",
               color=AI_COLOR, alpha=ALPHA, edgecolor=EDGE, linewidth=0.6)
p_ctrl = ax5.bar(x_paradox + 0.2, ctrl_scores, 0.4, label="Control",
                 color=CTRL_COLOR, alpha=ALPHA, edgecolor=EDGE, linewidth=0.6)

ax5.set_title('THE AI PARADOX — "Feels Easier, Thinks Harder"', fontsize=TITLE_SIZE, pad=12)
ax5.set_xlabel("Metrics", fontsize=LABEL_SIZE)
ax5.set_ylabel("Scores", fontsize=LABEL_SIZE)
ax5.set_xticks(x_paradox)
ax5.set_xticklabels(categories, fontsize=TICK_SIZE)
ax5.grid(True, axis="y", alpha=0.25)
ax5.legend(fontsize=10, frameon=False, loc="upper left")

# Labels
add_bar_labels(ax5, p_ai, vpad=0.08)
add_bar_labels(ax5, p_ctrl, vpad=0.08)

# Arrows/notes: effort ↓ with AI (green), load ↑ with AI (red)
add_paradox_arrows(ax5, 0, ai_scores[0], ctrl_scores[0], "AI reduces effort", color="#2E7D32")  # green
add_paradox_arrows(ax5, 1, ai_scores[1], ctrl_scores[1], "AI increases load", color="#C62828")  # red

# 6) Summary box
ax6 = fig.add_subplot(gs[2, :2])
ax6.axis("off")

summary = (
    "KEY FINDINGS\n"
    f"• PERCEIVED EFFORT (lower is better): AI {avg_ai_effort:.2f} vs Control {avg_ctrl_effort:.2f}  "
    f"→ Δ = −{effort_reduction:.2f} (−{effort_reduction/avg_ctrl_effort*100:.1f}%)\n"
    f"• COGNITIVE LOAD (higher = more overwhelmed): AI {avg_ai_load:.2f} vs Control {avg_ctrl_load:.2f}  "
    f"→ Δ = +{load_increase:.2f} (+{load_increase/avg_ctrl_load*100:.1f}%)\n\n"
    "Interpretation: AI lowers *perceived effort* but raises *mental workload* due to verification, "
    "integration, and sense-making of AI outputs."
)

ax6.text(
    0.02, 0.95, summary, va="top", ha="left",
    fontsize=12, family="DejaVu Sans Mono",
    bbox=dict(boxstyle="round,pad=0.6", facecolor="#EEF7F6", edgecolor="#A7E0D8", alpha=0.9)
)

# Suptitle
fig.suptitle("AI PARADOX: Decreases Perceived Effort, Increases Cognitive Load",
             fontsize=18, fontweight="bold", y=0.995)

plt.show()

# -------------------------
# Console deep-dive
# -------------------------
print("="*84)
print("DEEP DIVE (per country)")
print("="*84)
for i, c in enumerate(countries):
    eff_drop = control_perceived_effort[i] - ai_perceived_effort[i]
    load_up  = ai_cognitive_load[i] - control_cognitive_load[i]
    print(f"• {c:10s}: Effort ↓ {eff_drop:.1f} pts | Load ↑ {load_up:.1f} pts")
print("="*84)


```
